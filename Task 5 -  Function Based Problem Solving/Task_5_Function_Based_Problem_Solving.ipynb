```python
#3.Agriculture – Rainfall Adequacy Checker
def rainfall_checker(rainfall_list, required_level):
    total = 0
    for rain in rainfall_list:
        total += rain
    average = total / len(rainfall_list)
    print("Average Rainfall:", int(average))
    if average >= required_level:
        print("Rainfall Status: Adequate Rainfall")
    else:
        print("Rainfall Status: Inadequate Rainfall")
seasonal_data = [70, 75, 80, 65, 70]
required = 70
rainfall_checker(seasonal_data, required)
```

```python
#4.Social Media – Duplicate Account Detection
def check_duplicate_accounts(usernames):
    unique_users = set(usernames)
    if len(unique_users) != len(usernames):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")
accounts = ["gopi123", "dharani", "user01", "gopi123"]
check_duplicate_accounts(accounts)
```

```python
#6.Agriculture – Premium Crop Price Filter
def filter_premium_crops(prices):
    premium=[]
    for price in prices:
        if price>2000:
            premium.append(price)
    print("Premium Crops:",premium)

crop_prices=[1500,2500,1800,3200,2000]
filter_premium_crops(crop_prices)
```

```python
#7.System Monitoring – Application Health Checker
def check_system_health(errors):
    print("Error Count:"+str(errors))
    if errors==0:
        print("System Status:Healthy")
    elif errors<=5:
        print("System Status:Minor Issues")
    else:
        print("System Status:Critical Issues")
error_count=7
check_system_health(error_count)
```

```python
#8.Banking – Daily Transaction Limit Checker
def check_transaction(amount):
    limit=50000
    if amount<=limit:
        status="Approved"
    else:
        status="Rejected"
    print("Transaction Amount:",amount)
    print("Transaction Status:",status)

check_transaction(60000)
```

```python
#9.E-Learning – Student Attendance Eligibility System
def check_eligibility(attendance_list):
    total_classes=len(attendance_list)
    attended_classes=0
    for status in attendance_list:
        if status==1:
            attended_classes+=1
    percentage=(attended_classes/total_classes)*100
    if percentage>=75:
        eligibility="Eligible"
    else:
        eligibility="Not Eligible"
    print("Attendance Percentage:",percentage)
    print("Exam Eligibility:",eligibility)
attendance=[1,1,1,1,0]
check_eligibility(attendance)
```

```python
#10.Smart Electricity Bill Analyzer
def calculate_bill(units):
    bill=0
    if units<=100:
        bill=units*3
    elif units<=200:
        bill=(100*3)+((units-100)*5)
    else:
        bill=(100*3)+(100*5)+((units-200)*7)

    if bill<500:
        usage="Low Usage"
    elif bill<=1500:
        usage="Moderate Usage"
    else:
        usage="High Usage"
    return bill,usage
bill_amount,usage_status=calculate_bill(250)
print("Total Bill:",bill_amount)
print("Usage Level:",usage_status)
```

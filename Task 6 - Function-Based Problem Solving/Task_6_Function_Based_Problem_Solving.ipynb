```python
#1.Smart Parking Lot Management System
def parking_system(capacity, logs):
    parked=0
    peak=0
    for log in logs:
        if log=="IN":
            parked+=1
            if parked>peak:
                peak=parked
        elif log=="OUT" and parked>0:
            parked-=1
    print("Currently Parked Vehicles:",parked)
    if parked>capacity:
        print("Parking Status: Exceeded Capacity")
    else:
        print("Parking Status: Available")
    print("Peak Usage:",peak)
parking_system(50,["IN","IN","IN","OUT","IN","IN","OUT"])
```

```python
#2.Online Food Delivery Time Estimator
def delivery_time(distance,traffic,weather):
    eta=distance*5
    if traffic=="High":
        eta+=15
    elif traffic=="Medium":
        eta+=8
    if weather=="Rainy":
        eta+=10
    print("Estimated Delivery Time:",eta,"minutes")
delivery_time(8,"High","Rainy")
```

```python
#3.Movie Theatre Seat Occupancy Analyzer
def theatre_status(total_seats,booked):
    occupancy=(len(booked)/total_seats)*100
    print("Occupancy:",int(occupancy),"%")
    if occupancy==100:
        print("Show Status: Housefull")
    elif occupancy>=75:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Available")
theatre_status(200,[1]*150)
```

```python
#4:Cloud Server Load Classification System
def server_load(cpu_readings):
    avg=sum(cpu_readings)/len(cpu_readings)
    print("Average CPU Load:",int(avg),"%")
    if avg<50:
        print("Server Status: Normal")
    elif avg<=80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")
server_load([45,60,70,85,90])
```

```python
#5.Smart Classroom Resource Usage Monitor
def classroom_monitor(resources):
    overused=[]
    for r in resources:
        if resources[r]>8:
            overused.append(r)
    if overused:
        print("Overused Resources:",",".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")
classroom_monitor({"Projector":6,"AC":9,"Lights":4})
```

```python
#6.Online Event Registration Capacity Controller
def event_registration(capacity,registrations):
    if registrations>capacity:
        confirmed=capacity
        waitlist=registrations-capacity
        status="Closed"
    else:
        confirmed=registrations
        waitlist=0
        status="Open"
    print("Confirmed Registrations:",confirmed)
    print("Waitlisted Users:",waitlist)
    print("Registration Status:",status)
event_registration(100,105)
```
